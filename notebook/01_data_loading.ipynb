```python
# Seasons kitne hain?
print('PSL Seasons in dataset:')
print(sorted(df['season'].unique()))
print()
print('Matches per season:')
print(df.groupby('season')['match_id'].nunique().to_string())
```

```python
# Teams
print('PSL Teams:')
for team in sorted(df['batting_team'].unique()):
    print(f'  - {team}')
```

```python
# Venues
print('Venues:')
for v in sorted(df['venue'].unique()):
    print(f'  - {v}')
```

```python
# Missing values check
print('Missing Values per Column:')
print('='*40)
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct})
print(missing_df[missing_df['Missing Count'] > 0].to_string())
```
